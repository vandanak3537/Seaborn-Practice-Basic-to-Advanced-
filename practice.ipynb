```python
#Importing Seaborn
import seaborn as sns
import matplotlib.pyplot as plt
import  pandas as pd
```

```python
#creating a small dataset
df=pd.DataFrame({
    "Student":["A","B","C","D","E","F"],
    "Study_Hours":[2,4,3,5,6,4],
   "Marks":[55,68,60,82,90,72],
    "Department":["CSE","ECE","CSE","AIML","AIML","ECE"],
    "Gender":["F","M","M","M","F","F"]
})

df
```

```python
sns.scatterplot(data=df,x="Study_Hours",y="Marks")
plt.show()
```

```python
#hue
sns.scatterplot(data=df,x="Study_Hours",y="Marks",hue="Department")
plt.show()
```

```python
#style
sns.scatterplot(data=df,x="Study_Hours",y="Marks",hue="Department",style="Department")
plt.show()
```

```python
#Size
sns.scatterplot(data=df,x="Study_Hours",y="Marks",size="Marks")
plt.show()
```

```python
#Line plot
sales=pd.DataFrame({
    "Month":["jan","feb","mar","april","may","jun"],
    "Sales":[100,150,120,180,200,220]
})
```

```python
sns.lineplot(data=sales,x="Month",y="Sales")
plt.show()
```

```python
#Adding markers
sns.lineplot(data=sales,x="Month",y="Sales",marker="o")
```

```python
import pandas as pd
df = pd.DataFrame({
    "Marks": [45, 50, 55, 60, 65, 70, 72, 78, 82, 88],
    "Department": ["A", "A", "A", "A", "A",
                   "B", "B", "B", "B", "B"]
})
```

```python
#distplot
sns.histplot(data=df, x="Marks", kde=True)
plt.show()
```

```python
#pairplot()
df = pd.DataFrame({
    "Study_Hours": [2, 3, 4, 5, 6],
    "Marks": [50, 55, 65, 72, 85],
    "Attendance": [60, 65, 70, 80, 90]
})

sns.pairplot(df)
plt.show()
```

```python
#pair plot with hue
df["Gender"] = ["M", "F", "M", "F", "M"]
sns.pairplot(df,hue="Gender")
```

```python
#joint plot()
sns.jointplot(data=df, x="Study_Hours", y="Marks")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.DataFrame({
    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 3, 5, 7],
    "Marks": [45, 50, 60, 65, 70, 78, 88, 55, 68, 82],
    "Attendance": [60, 65, 70, 75, 80, 85, 90, 68, 78, 88],
    "Sleep_Hours": [8, 7, 7, 6, 6, 6, 5, 8, 7, 6]
})
```

```python
#corelation
corr=df.corr(numeric_only=True)
print(corr)
```

```python
#corelation heatmap
sns.heatmap(corr,annot=True, cmap="coolwarm")
plt.show()
```

```python
#catplot()
sns.catplot(data=df,x="Study_Hours",y="Marks",kind="box")
plt.show()
```

```python
#relplot()
sns.relplot(
    data=df,
    x="Study_Hours",
    y="Marks",
    hue="Department"
)

plt.show()
```

```python
#implot()
sns.lmplot(
    data=df,
    x="Study_Hours",
    y="Marks"
)

plt.show()
```
